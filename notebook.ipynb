```python
# Definição dos Fatos (Estruturas de Dados)
class Usuario(Fact):
    """Dados fornecidos pelo usuário"""
    objetivo = Field(str)          # "hipertrofia" ou "emagrecimento"
    biotipo = Field(str)           # "ectomorfo" ou "endomorfo"
    restricao = Field(str)         # "nenhuma" ou "intolerante_lactose"
    experiencia = Field(str)       # "iniciante" ou "avancado"

class PerfilMetabolico(Fact):
    """Fato intermediário (Nível 1)"""
    tipo = Field(str)              # "gasto_rapido" ou "acumulo_gordura"

class Planejamento(Fact):
    """Fato intermediário (Nível 2)"""
    dieta = Field(str)             # "hipercalorica" ou "hipocalorica"
    treino_base = Field(str)        # "forca_hipertrofia" ou "deficit_calorico_com_cardio"

class RecomendacaoFinal(Fact):
    """Fato final (Nível 3)"""
    dieta_detalhe = Field(str)
    treino_detalhe = Field(str)
    suplementacao = Field(str)
```

```python
# Sistema de Regras
class MotorFitTech(KnowledgeEngine):

    @Rule(Usuario(objetivo="emagrecimento", biotipo="ectomorfo"), salience=10)
    def regra_alerta_incompatibilidade(self):
        """Regra 1: Alerta de Incompatibilidade Biológica"""
        print("[TRACE - Regra 1 disparada] Conflito biológico detectado: Ectomorfo buscando emagrecimento radical.")
        self.declare(PerfilMetabolico(tipo="gasto_rapido"))
        print("  -> Decisão: Forçado PerfilMetabolico = gasto_rapido para evitar catabolismo.")

    # NÍVEL 1

    @Rule(Usuario(objetivo="hipertrofia", biotipo="ectomorfo"))
    def regra_perfil_ectomorfo_hipertrofia(self):
        """Regra 2: Perfil Ectomorfo focado em Ganho"""
        print("[TRACE - Regra 2 disparada] Usuário é Ectomorfo focado em Hipertrofia.")
        self.declare(PerfilMetabolico(tipo="gasto_rapido"))
        print("  -> Decisão: Declarado PerfilMetabolico = gasto_rapido.")

    @Rule(Usuario(objetivo="emagrecimento", biotipo="endomorfo"))
    def regra_perfil_endomorfo_emagrecimento(self):
        """Regra 3: Perfil Endomorfo focado em Perda"""
        print("[TRACE - Regra 3 disparada] Usuário é Endomorfo focado em Emagrecimento.")
        self.declare(PerfilMetabolico(tipo="acumulo_gordura"))
        print("  -> Decisão: Declarado PerfilMetabolico = acumulo_gordura.")

    # NÍVEL 2

    @Rule(PerfilMetabolico(tipo="gasto_rapido"), Usuario(restricao="nenhuma"))
    def regra_plano_hipercalorico_padrao(self):
        """Regra 4: Dieta Hipercalórica Padrão"""
        print("[TRACE - Regra 4 disparada] Perfil gasto_rapido sem restrições alimentares.")
        self.declare(Planejamento(dieta="hipercalorica_padrao", treino_base="forca_hipertrofia"))
        print("  -> Decisão: Definido Planejamento com Dieta Hipercalórica Padrão.")

    @Rule(PerfilMetabolico(tipo="gasto_rapido"), Usuario(restricao="intolerante_lactose"))
    def regra_plano_hipercalorico_restrito(self):
        """Regra 5: Dieta Hipercalórica sem Lactose"""
        print("[TRACE - Regra 5 disparada] Perfil gasto_rapido COM intolerância à lactose.")
        self.declare(Planejamento(dieta="hipercalorica_sem_lactose", treino_base="forca_hipertrofia"))
        print("  -> Decisão: Definido Planejamento com Dieta Hipercalórica Sem Lactose.")

    @Rule(PerfilMetabolico(tipo="acumulo_gordura"))
    def regra_plano_hipocalorico(self):
        """Regra 6: Dieta Hipocalórica e Cardio"""
        print("[TRACE - Regra 6 disparada] Perfil acumulo_gordura necessita de déficit.")
        self.declare(Planejamento(dieta="hipocalorica_restrita", treino_base="deficit_calorico_com_cardio"))
        print("  -> Decisão: Definido Planejamento com Dieta Hipocalórica.")

    # NÍVEL 3
 
    @Rule(Planejamento(dieta="hipercalorica_padrao", treino_base="forca_hipertrofia"), Usuario(experiencia="iniciante"))
    def regra_final_ectomorfo_iniciante(self):
        """Regra 7: Finalização Ectomorfo Iniciante"""
        print("[TRACE - Regra 7 disparada] Customizando para Iniciante em Hipertrofia Padrão.")
        self.declare(RecomendacaoFinal(
            dieta_detalhe="Superávit de 500kcal com ingestão alta de leite integral e derivados.",
            treino_detalhe="Treino ABC 3x na semana, foco em exercícios compostos, descanso longo.",
            suplementacao="Hipercalórico e Creatina."
        ))

    @Rule(Planejamento(dieta="hipercalorica_sem_lactose", treino_base="forca_hipertrofia"), Usuario(experiencia="iniciante"))
    def regra_final_ectomorfo_restrito_iniciante(self):
        """Regra 8: Finalização Ectomorfo com Restrição Iniciante"""
        print("[TRACE - Regra 8 disparada] Customizando para Iniciante em Hipertrofia Sem Lactose.")
        self.declare(RecomendacaoFinal(
            dieta_detalhe="Superávit de 500kcal utilizando leite de amêndoas, aveia e carnes magras.",
            treino_detalhe="Treino ABC 3x na semana, foco em execução correta.",
            suplementacao="Proteína isolada da carne (Beef Protein) e Creatina."
        ))

    @Rule(Planejamento(dieta="hipocalorica_restrita", treino_base="deficit_calorico_com_cardio"), Usuario(experiencia="iniciante"))
    def regra_final_endomorfo_iniciante(self):
        """Regra 9: Finalização Endomorfo Iniciante"""
        print("[TRACE - Regra 9 disparada] Customizando para Iniciante focado em queima de gordura.")
        self.declare(RecomendacaoFinal(
            dieta_detalhe="Déficit de 400kcal. Alto teor de proteínas, vegetais fibrosos e redução de carbo simples.",
            treino_detalhe="Treino AB 4x na semana + 20 min de cardio moderado pós-treino.",
            suplementacao="Whey Protein e Termogênico leve (Café)."
        ))

    @Rule(Planejamento(treino_base="deficit_calorico_com_cardio"), Usuario(experiencia="avancado"))
    def regra_final_endomorfo_avancado(self):
        """Regra 10: Finalização Endomorfo Avançado (Ajuste de intensidade)"""
        print("[TRACE - Regra 10 disparada] Customizando para Avançado focado em queima de gordura.")
        self.declare(RecomendacaoFinal(
            dieta_detalhe="Déficit de 600kcal com estratégia de ciclagem de carboidratos.",
            treino_detalhe="Treino ABCDE de alta intensidade (Drop-sets) + 40 min HIIT em dias alternados.",
            suplementacao="Whey Isolado, Termogênico ultra e BCAA."
        ))

    # Regra de exibição do output final
    @Rule(RecomendacaoFinal(dieta_detalhe=MATCH.d, treino_detalhe=MATCH.t, suplementacao=MATCH.s), salience=-10)
    def imprimir_resultado(self, d, t, s):
        print("\n" + "="*50)
        print("🎯 RECOMENDAÇÕES GERADAS PELO SISTEMA ESPECIALISTA")
        print("="*50)
        print(f"🥗 DIETA:       {d}")
        print(f"🏋️‍♂️ TREINO:      {t}")
        print(f"💊 SUPLEMENTOS: {s}")
        print("="*50 + "\n")
```

```python
def executar_teste(titulo, dados_usuario):
    print(f"\n🚀 {titulo}")
    engine = MotorFitTech()
    engine.reset()
    engine.declare(Usuario(**dados_usuario))
    engine.run()

```

```python
# Caso 1: Ectomorfo, Iniciante, sem restrições procurando Ganhar Massa
executar_teste("Caso de Teste 1: Ectomorfo Iniciante Padrão", {
    "objetivo": "hipertrofia",
    "biotipo": "ectomorfo",
    "restricao": "nenhuma",
    "experiencia": "iniciante"
})

```

```python
# Caso 2: Ectomorfo, Iniciante com Intolerância a Lactose
executar_teste("Caso de Teste 2: Ectomorfo com Restrição Alimentar", {
    "objetivo": "hipertrofia",
    "biotipo": "ectomorfo",
    "restricao": "intolerante_lactose",
    "experiencia": "iniciante"
})

```
